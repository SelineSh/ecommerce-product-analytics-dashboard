### Order Value Distribution

We grouped payment records by order ID to calculate the total payment value for each order. This step is necessary because some orders may have multiple payment records.

The average order value is 160.99, while the median order value is 105.29. This suggests that the order value distribution is right-skewed, with a small number of high-value orders increasing the average.

Use `product_id` to perform a left join between the order item details `items` and the product table `products`.


```python
items_products = items.merge(
    products,
    on="product_id",
    how="left"
)

items_products.head()
```

Group by product category, sum the `price` values in the order item details, and sort them from highest to lowest.


```python
category_sales = (
    items_products
    .groupby("product_category_name")["price"]
    .sum()
    .sort_values(ascending=False)
)

category_sales.head(10)
```

Create a bar chart showing the top 10 product categories from `category_sales`, and add a chart title and vertical axis labels.


```python
import matplotlib.pyplot as plt

category_sales.head(10).plot(kind="bar")

plt.title("Top 10 Product Categories by Revenue")
plt.ylabel("Revenue")
plt.show()
```
